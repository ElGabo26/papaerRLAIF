```python
import pandas as pd
import os
```

```python
route="./"
bases=os.listdir(route)
bases=[x for x in  bases if '.csv' in x]
bases_result=[pd.read_csv(x, index_col= 0) for  x in bases]
idx=0
for x in bases_result:
    x['id_prompt']=range(len(x))
    x.to_csv(f"finalPromptBases/{bases[idx]}")
    idx+=1
```
